```python
import pandas as pd

df = pd.read_csv(r"C:\Users\raiva\OneDrive\Desktop\BBDU Traning\data.csv\data.csv", encoding='latin1')
df["IsCancelled"] = df["InvoiceNo"].astype(str).str.startswith("C")
df["IsCancelled"]
```

```python
df_clean = df[
    (~df["IsCancelled"]) &
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)

].copy()
```

```python
df_clean
```

```python
import numpy as np
df_clean["TotalPrice"] = np.multiply(df_clean["Quantity"], df_clean["UnitPrice"])
df_clean["TotalPrice"]
```

```python
print("Step 3: Data Cleaned")
print("Originals Rows:", df.shape[0], "Cleaned Rows:", df_clean.shape[0])
print("Cancelled Ordersremoved:", df["IsCancelled"].sum())
print(df_clean[["InvoiceNo", "Quantity", "UnitPrice", "TotalPrice","InvoiceDate"]])
print("\n")
```

```python
print(df_clean["InvoiceDate"].dtype)
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
```

```python
df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["Day"] = df_clean["InvoiceDate"].dt.day
df_clean["DayOfWeek"] = df_clean["InvoiceDate"].dt.dayofweek    
df_clean["Hour"] = df_clean["InvoiceDate"].dt.hour
df_clean["DayOfWeekName"] = df_clean["InvoiceDate"].dt.day_name()
df_clean
```
